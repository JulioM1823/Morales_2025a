```python
# Load the local pipeline configuration and standalone dispersion-comparison helpers.

from copy import deepcopy
import importlib.util
import json
from pathlib import Path
import sys

from IPython.display import Image, display


def load_local_module(name, path):
    '''
    Purpose
    -------
    Register and execute a local module from its explicit source file.

    Inputs
    ------
    name: str
        Import name registered in sys.modules.

    path: str or Path
        Python source file to execute.

    Outputs
    -------
    module: types.ModuleType
        Executed module object; import failures propagate after registration.
    '''

    spec = importlib.util.spec_from_file_location(name, path)
    module = importlib.util.module_from_spec(spec)
    sys.modules[name] = module
    spec.loader.exec_module(module)
    return module


candidates = [Path.cwd(), Path.cwd() / 'Code' / 'Time-Distance']
TD_DIR = next((path.resolve() for path in candidates if (path / 'config.py').exists()), None)
if TD_DIR is None:
    raise FileNotFoundError('Run this notebook from its directory or from the Morales_2025a project root.')

config_module = load_local_module('komega_demo_config', TD_DIR / 'config.py')
dispersion = load_local_module('komega_demo_dispersion', TD_DIR / 'dispersion_comparison.py')
config = deepcopy(config_module.config)
```

```python
# Change DATASET_PATH to the k-omega FITS dataset to compare.
# All six theoretical models are calculated and displayed together.
# The FITS file does not need simulation-layer metadata.
# Missing a, theta, phi, and dz values come from config['komega_comparison']['standalone_pair'].
# Missing FITS axis WCS comes from config['komega_comparison']['standalone_sampling'].
configured_dataset = str(config['data'].get('komega_outfile', '')).strip()
DATASET_PATH = Path(configured_dataset).expanduser() if configured_dataset else Path('/Users/juliomorales/Library/CloudStorage/GoogleDrive-juliomorales1823@gmail.com/My Drive/Graduate School/Research/Projects/Helioseismology/Morales_2025a/Data/Time-Distance/filter_10/simulations/nonmagneto/horizontal/10G/komega/annulus/v3/hx_10g_v3_440km_660km_coherence.fits')

# Output folder; plotting style comes from config['komega_comparison']['plot_style'].
OUTPUT_DIR = Path.cwd() / 'notebook_komega_plots'
```

```python
# Generate the configured standalone comparison and display its saved image.

dataset = dispersion.load_komega_fits_dataset(DATASET_PATH, config)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

comparison_file = OUTPUT_DIR / f'{DATASET_PATH.stem}_seven_row_comparison.png'
comparison = dispersion.generate_standalone_komega_comparison(
    dataset,
    config,
    comparison_file,
)

display(Image(filename=str(comparison_file)))

print('Dataset:', dataset['dataset_file'])
print('Seven-row comparison:', comparison_file)
print('Rows:')
for number, row in enumerate(comparison['manifest']['rows'], start=1):
    print(f"  {number}. {row['label']}")
print('Height-pair context:')
print(json.dumps(comparison['manifest']['pair_context'], indent=2))
print('Branches and parameters supplied to every model:')
for model in comparison['manifest']['models']:
    print(f"{model['label']}: {model['kz_branches_rendered']} branches")
    print(json.dumps(model['params_supplied'], indent=2))
```
